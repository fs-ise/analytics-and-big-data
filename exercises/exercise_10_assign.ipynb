# Exercise 10: Governance of data analytics

## Purpose

In this exercise, you will analyze how analytics contributes to
strategic objectives in a real company. Your task is to identify where
analytics enters concrete business processes, which decisions it
supports, what data and systems are needed, and how the organization
learns from outcomes.

You will develop an **Analytics-Enabled Strategy Map** for one selected
company.

## Learning objectives

After completing this exercise, you should be able to:

1.  Connect a strategic objective to a North Star metric and related
    KPIs.
2.  Identify business processes in which analytics creates value.
3.  Explain which decisions are supported by analytics.
4.  Describe the data sources and systems needed for analytics-enabled
    work.
5.  Reflect on learning loops, governance mechanisms, and regulatory or
    ethical boundaries.

## Step 1: Select your case

Select **one** company from the following list:

| Case        | Industry                              |
|-------------|---------------------------------------|
| Netflix     | Streaming / entertainment             |
| Meta        | Social media / digital platforms      |
| X / Twitter | Social media / digital platforms      |
| Siemens     | Manufacturing / industrial technology |
| Capital One | Banking / financial technology        |

You may propose another case, but it must have enough public material to
analyze strategy, processes, analytics, data, systems, and governance.

## Step 2: Familiarize yourself with the company

Start by building a basic understanding of the company:

- What is the company’s business model?
- Who are its main customers or users?
- What are its main products, services, or platforms?
- What strategic objective seems especially important for the company?
- What could be a plausible North Star metric or key performance
  indicator?

Do not assume that the official North Star metric is publicly available.
If necessary, infer a plausible metric and justify your reasoning.

## Step 3: Conduct complementary web research

Use public materials such as annual reports, investor presentations,
engineering blogs, research blogs, transparency reports, product
documentation, regulatory documents, high-quality journalism, or
academic case studies.

Use at least **three credible sources**. These can be company sources,
external sources, or a combination of both.

For each source, ask:

- What strategic objective does this source reveal?
- Which business processes are described?
- Which decisions are supported by analytics?
- Which data sources or systems are mentioned?
- Are there signs of experimentation, feedback, monitoring, governance,
  or regulation?

## Step 4: Build an Analytics-Enabled Strategy Map

Fill out the following table for your selected case.

| Layer | Guiding question | Notes |
|------------------------|------------------------|------------------------|
| Strategic objective / North Star metric | What does the company want to achieve, and what overarching metric could indicate success? |  |
| Operational KPIs | Which more specific KPIs help track progress toward the strategic objective? |  |
| Business process | In which process is value created? |  |
| Decision point | What decision must improve? |  |
| Data sources | What data is used to support this decision? |  |
| Systems / applications | Which systems, platforms, or tools are needed? |  |
| Action | What changes in practice because of analytics? |  |
| Outcome | What business, user, or operational outcome is expected? |  |
| Feedback / learning | How does the company learn from results? |  |
| Governance / boundaries | What legal, ethical, technical, or organizational limits matter? |  |

> **Notes**
>
> Focus on **specific analytics-enabled decisions**, not broad
> statements such as “the company uses AI.” The goal is to understand
> how analytics connects strategic objectives to concrete processes,
> decisions, actions, and outcomes.
>
> Some information may not be publicly available. In such cases, make a
> careful, evidence-based inference and clearly indicate where you are
> interpreting rather than reporting.
